```python

import numpy as np
import pandas as pd
import joblib

# Load model and preprocessors
model = joblib.load("randomforest_model.pkl")   # or logistic_model.pkl / knn_model.pkl
scaler = joblib.load("scaler.pkl")
imputer = joblib.load("imputer.pkl")
label_encoder = joblib.load("label_encoder.pkl")

# Try to get feature names from model if available
if hasattr(model, "feature_names_in_"):
    trained_features = list(model.feature_names_in_)
else:
    # fallback if feature_names_in_ not stored
    print("⚠️ Model doesn't store feature names; please ensure correct order manually.")
    trained_features = ["pH","humidity","temperature","moisture","ldr_status"
    ]

print("Model expects features:\n", trained_features)

# Example new input (replace with actual sensor data)
input_data = {
    "pH": 27.52,
    "humidity": 22.24,
    "temperature": 21.90,
    "moisture": 55.29,
    "ldr_status": 556.17
}

# Convert to DataFrame
input_df = pd.DataFrame([input_data])

# --- Match training feature names ---
# Add any missing features as NaN
for f in trained_features:
    if f not in input_df.columns:
        input_df[f] = np.nan

# Reorder columns to match training
input_df = input_df[trained_features]

# Apply preprocessing (same as training)
X_imputed = imputer.transform(input_df)
X_scaled = scaler.transform(X_imputed)

# Predict
pred_encoded = model.predict(X_scaled)[0]
pred_label = label_encoder.inverse_transform([pred_encoded])[0]

print("\n🌱 Predicted Pump Status:", pred_label)
```
